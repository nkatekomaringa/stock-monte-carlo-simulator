# Monte Carlo Stock Price Forecast & Risk Modeling
Interactive notebook executing Geometric Brownian Motion simulations and risk metrics.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from src.data_loader import MarketDataLoader
from src.simulation import MonteCarloSimulator
from src.metrics import calculate_risk_metrics
```

```python
# Load AAPL market data
loader = MarketDataLoader('AAPL', period='2y')
prices = loader.fetch_data()
log_returns = loader.get_log_returns(prices)
last_price = float(prices.iloc[-1])

print(f'Last Close Price: ${last_price:.2f}')
```

```python
# Run Simulation
sim = MonteCarloSimulator(forecast_days=252, num_simulations=1000, seed=42)
paths = sim.run_simulation(last_price, log_returns)

plt.figure(figsize=(10, 5))
plt.plot(paths[:, :100], opacity=0.3)
plt.title('Monte Carlo Simulated Trajectories (AAPL - 1 Year)')
plt.xlabel('Trading Days')
plt.ylabel('Stock Price ($)')
plt.grid(True)
plt.show()
```
